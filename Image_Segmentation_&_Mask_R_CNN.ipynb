**1. What is TensorFlow Object Detection API (TFOD2) and what are its primary components?**

The TensorFlow Object Detection API (TFOD2) is an open-source framework built on TensorFlow 2.x that simplifies the development, training, and deployment of object detection and instance segmentation models. It provides pre-trained models, training pipelines, and evaluation tools for tasks such as bounding box detection and mask prediction.

**Primary Components of TFOD2:**

**1. Model Zoo**

Collection of pre-trained models such as SSD, Faster R-CNN, EfficientDet, and Mask R-CNN.

**2. Pipeline Configuration Files**

.config files defining model architecture, training parameters, dataset paths, and optimization settings.

**3. Dataset Utilities**

Tools for converting datasets into TFRecord format.

**4. Training Scripts**

Scripts like model_main_tf2.py for training and fine-tuning models.

**5. Evaluation Tools**

Metrics such as mAP, precision, recall, and IoU.

**6. Inference Utilities**

Scripts and functions to perform prediction on images and videos.

**3.  Explain the Mask R-CNN architecture. How does it extend Faster R-CNN?**

Mask R-CNN is an extension of Faster R-CNN that performs instance segmentation by adding a parallel branch for predicting object masks.

**Architecture Components:**

1. Backbone Network

ResNet + Feature Pyramid Network (FPN)

2. Region Proposal Network (RPN)

Generates region proposals

3. RoIAlign

Improves spatial alignment compared to RoIPool

4. Three Parallel Outputs:

- Bounding box regression

- Class prediction

- Pixel-wise mask prediction

**Extension Over Faster R-CNN:**

Faster R-CNN → bounding boxes + classes

Mask R-CNN → bounding boxes + classes + segmentation masks

**4.  Describe the purpose of masks in image segmentation. How are they used during training and inference?**

Masks represent pixel-level object boundaries, allowing precise shape detection.

During Training:

Ground truth masks are compared with predicted masks.

Loss functions (binary cross-entropy) are applied to each pixel.

During Inference:

Model predicts masks for detected objects.

Masks are overlaid on images for visualization or further analysis.

**8. Load a TFRecord dataset and visualize annotation masks.**

```python
import tensorflow as tf
import matplotlib.pyplot as plt

def parse_tfrecord(example):
    features = {
        'image/encoded': tf.io.FixedLenFeature([], tf.string),
        'image/object/mask': tf.io.VarLenFeature(tf.string),
    }
    parsed = tf.io.parse_single_example(example, features)
    image = tf.image.decode_jpeg(parsed['image/encoded'])
    return image

dataset = tf.data.TFRecordDataset("dataset.tfrecord")
dataset = dataset.map(parse_tfrecord)

for image in dataset.take(1):
    plt.imshow(image)
    plt.title("Image with Annotation Mask")
    plt.axis("off")
```
