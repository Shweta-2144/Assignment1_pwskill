**1. What is hypothesis testing in statistics?**

Hypothesis testing is a statistical method used to make decisions or draw conclusions about a population based on sample data.

It helps determine whether there is enough evidence to support a particular claim or assumption about a population parameter.

Steps in Hypothesis Testing:

Define the null hypothesis (H0)

Define the alternative hypothesis (H1)

Choose significance level (α)

Calculate test statistic

Find p-value or critical value

Accept or reject the null hypothesis

Applications:

- Medical research
- Business analysis
- Quality control
- Machine learning evaluation

**3. Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.**

The significance level, denoted by α, is the probability of rejecting the null hypothesis when it is actually true.

Common significance levels:

0.05 (5%)

0.01 (1%)

Role of Significance Level:

- Determines the threshold for decision-making.
- If p-value < significance level:
- Reject the null hypothesis.
- If p-value > significance level:
- Fail to reject the null hypothesis.

Example:

If:

p-value=0.03

and:

α=0.05

Then:

0.03<0.05

So, reject the null hypothesis.

**7. Implement hypothesis testing using Z-statistics for the sample dataset.**

```python
import numpy as np
from scipy.stats import norm

sample_data = [
49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]

# Parameters
population_mean = 50

# Calculations
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data)
n = len(sample_data)

# Z-score
z = (sample_mean - population_mean) / (sample_std / np.sqrt(n))

# P-value
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Sample Mean:", sample_mean)
print("Z-score:", z)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

**8. Simulate data from a normal distribution and calculate 95% confidence interval.**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Generate normal distribution data
data = np.random.normal(loc=50, scale=5, size=100)

# Mean and standard error
mean = np.mean(data)
std_error = stats.sem(data)

# 95% confidence interval
confidence_interval = stats.t.interval(
    0.95,
    len(data)-1,
    loc=mean,
    scale=std_error
)

print("Mean:", mean)
print("95% Confidence Interval:", confidence_interval)

# Plot histogram
plt.hist(data, bins=15)

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

**9. Calculate Z-scores and visualize standardized data using histogram.**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

# Sample dataset
data = [10, 12, 14, 15, 18, 20, 22, 25, 30]

# Calculate Z-scores
z_scores = zscore(data)

print("Z-scores:")
print(z_scores)

# Plot histogram
plt.hist(z_scores, bins=6)

plt.title("Histogram of Z-scores")
plt.xlabel("Z-score")
plt.ylabel("Frequency")

plt.show()
```
