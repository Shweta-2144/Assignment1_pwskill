**14. Write a program to reverse a string using a while loop.**


```python
# Program to reverse a string using a while loop

def reverse_string(input_string):
    reversed_string = ""
    index = len(input_string) - 1  # Start from the last character

    while index >= 0:
        reversed_string += input_string[index]  # Append each character from end
        index -= 1  # Move to the previous character

    return reversed_string

# Example usage
original_string = "Hello, World!"
reversed_result = reverse_string(original_string)
print("Original String:", original_string)
print("Reversed String:", reversed_result)
```

**15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.**

```python
# Factorial using while loop

num = int(input("Enter a number: "))

factorial = 1
i = 1

# Check if the number is negative
if num == 0:
    print("The factorial of 0 is 1.")
else:
    while i <= num:
        factorial *= i
        i += 1
    print(f"The factorial of {num} is {factorial}.")
```
