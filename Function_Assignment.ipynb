1. What is the difference between a function and a method in Python?

A function is a block of code that works independently, while a method is a function that belongs to an object (like a string or list).

A function is defined using def and called independently, whereas a method is called using a dot (.) on an object like a string or list.

6. Explain the concept of generators in Python and how they are defined.

A generator is a special type of iterator in Python, defined using a function with the yield keyword.
It generates values one at a time.

- Generators don’t store the entire sequence in memory.

- They pause after each yield and resume from where they left off.

- Useful for large data processing, streams, pipelines, etc.

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

for num in count_up_to(3):
    print(num)
```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

map() applies a function to each item in an iterable and returns a new iterable of results.
Example: map(lambda x: x*2, [1, 2, 3]) → [2, 4, 6]

filter() selects items from an iterable based on a condition (function returning True or False).
Example: filter(lambda x: x%2==0, [1, 2, 3, 4]) → [2, 4]

reduce() (from functools) reduces an iterable to a single value by applying a function cumulatively.
Example: reduce(lambda x, y: x + y, [1, 2, 3, 4]) → 10

```python
#map()

nums = [1, 2, 3]
result = list(map(lambda x: x * 2, nums))
print(result)
```

```python
#filter()

nums = [1, 2, 3]
result = filter(lambda x: x%2==0, nums)
print(result)
```

```python
next(result)
```

```python
#reduce

from functools import reduce

nums = [1, 2, 3]
result = reduce(lambda x, y: x + y, nums)
print(result)
```

1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.

```python
def sum_even(numbers):
    return sum(num for num in numbers if num % 2 == 0)


nums = [1, 2, 3, 4, 5, 6]
result = sum_even(nums)
print("Sum of even numbers:", result)
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(number):
    if number <= 1:
        return False
    if number <= 200:
        return True
    if number % 2 == 0 or number % 3 == 0:
        return False
print("Prime numbers from 1 to 200:")
for num in range(1, 201):
    if is_prime(num):
        print(num)
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
class Fibonacci:
    def __init__(self, n):
        self.a, self.b, self.count, self.n = 0, 1, 0, n

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return self.a

# Example
for i in Fibonacci(5):
    print(i)
```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def power_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

# Example
for num in power_of_two(4):
    print(num)
```
